## Basic Finance Applied Problems


1. A company has a current stock price of `$50` and is expected to pay a dividend of `$2` next year. If the required rate of return is 10%, what is the expected stock price in one year?

## Solution
To calculate the expected stock price in one year, we can use the Dividend Discount Model (DDM). The formula for the expected stock price is:
$$
P_1 = \frac{D_1}{r} + P_0
$$
Where:
- \(P_1\) = Expected stock price in one year
- \(D_1\) = Dividend expected next year
- \(r\) = Required rate of return
- \(P_0\) = Current stock price
Given:
- \(D_1 = 2\)
- \(r = 0.10\)
- \(P_0 = 50\)
Now, substituting the values into the formula:
$$
P_1 = \frac{2}{0.10} + 50 = 20 + 50 = 70
$$

## Problem 2

2. You are considering investing in a bond that has a face value of `$1,000`, a coupon rate of 5%, and matures in 10 years. If the current market interest rate is 6%, what is the price of the bond?

## Solution
To calculate the price of the bond, we can use the present value formula for bonds. The price of the bond is the sum of the present value of the coupon payments and the present value of the face value. The formula for the price of the bond is:
$$
P = \sum_{t=1}^{n} \frac{C}{(1+r)^t} + \frac{F}{(1+r)^n}
$$
Where:
- \(P\) = Price of the bond
- \(C\) = Coupon payment (Face value * Coupon rate)
- \(r\) = Market interest rate
- \(n\) = Number of years to maturity
- \(F\) = Face value of the bond


````python
# Annual coupon payment
face_value = 1000
coupon_rate = 0.05
coupon_payment = face_value * coupon_rate
print(f"Coupon Payment: ${coupon_payment}")
````

### Cash Flows
The bond will pay a coupon of `$50` each year for 10 years and a face value of `$1,000` at the end of 10 years. The cash flows are as follows:
- Year 1: `$50`
- Year 2: `$50`
- Year 3: `$50`
- Year 4: `$50`
- Year 5: `$50`
- Year 6: `$50`
- Year 7: `$50`
- Year 8: `$50`
- Year 9: `$50`
- Year 10: `$1,050` (coupon + face value)
```text
Today     Yr 1     Yr 2     Yr 3             Yr 10
  |--------|--------|--------|---- ... --------|
  ?        C        C        C                 C
                                             +1000
```

Bond Price = PV(annuity) + PV(face value)

You receive the same C every year for 10 years.

The PV of an ordinary annuity is:

$$
PV_{\text{coupons}} = C \left[\frac{1 - (1+r)^{-n}}{r}\right]
$$


````python
# Calculate the present value of the coupon payments
market_rate = 0.06
n_years = 10
pv_coupons = coupon_payment * (1 - (1 + market_rate) ** -n_years) / market_rate
print(f"Present Value of Coupons: ${pv_coupons:.2f}")
````

Present value of the $1,000

The $1,000 isn't an annuity. It's one future cash flow received ten years from now.

$$
PV_{\text{face value}} = \frac{F}{(1+r)^n} = \frac{1000}{(1+0.06)^{10}}
$$


````python
# Calculate the present value of the face value
pv_face_value = face_value / (1 + market_rate) ** n_years
print(f"Present Value of Face Value: ${pv_face_value:.2f}")
````

````python
# Calculate the total price of the bond
bond_price = pv_coupons + pv_face_value
print(f"Price of the Bond: ${bond_price:.2f}")
````

3. A company is considering a project that requires an initial investment of `$100,000` and is expected to generate cash flows of `$30,000` per year for 5 years. If the required rate of return is 8%, what is the net present value (NPV) of the project?

## Solution
To calculate the net present value (NPV) of the project, we can use the following formula:
$$
NPV = \sum_{t=1}^{n} \frac{CF_t}{(1+r)^t} - C_0
$$
Where:
- \(NPV\) = Net present value
- \(CF_t\) = Cash flow at time t
- \(r\) = Required rate of return
- \(n\) = Number of years
- \(C_0\) = Initial investment

````python
from financial_modeling.corporate_finance.utils import npv_series

# Given values
initial_investment = 100000
cash_flows = [-initial_investment] + [30_000] * 5  # Cash flows for 5 years
required_rate_of_return = 0.08
# Calculate the NPV of the project
project_npv = npv_series(required_rate_of_return, cash_flows)
print(f"Net Present Value (NPV) of the project: ${project_npv:.2f}")
````
