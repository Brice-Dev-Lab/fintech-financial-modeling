## Basic Finance Applied Problems


1. A company has a current stock price of `$50` and is expected to pay a dividend of `$2` next year. If the required rate of return is 10%, what is the expected stock price in one year?

2. You are considering investing in a bond that has a face value of `$1,000`, a coupon rate of 5%, and matures in 10 years. If the current market interest rate is 6%, what is the price of the bond?

3. A company is considering a project that requires an initial investment of `$100,000` and is expected to generate cash flows of `$30,000` per year for 5 years. If the required rate of return is 8%, what is the net present value (NPV) of the project?

4. A company has a debt-to-equity ratio of 0.5 and a total equity of `$200,000`. What is the total debt of the company?

5. A company has a current stock price of `$40` and is expected to pay a dividend of `$1.50` next year. If the required rate of return is 12%, what is the expected stock price in one year?

6. You just took a `$150,000` 30-year mortgage loan. Payments at the end of each year are flat (equal in every year) at an interest rate of 5%. Calculate the appropriate loan amortization schedule and the total interest paid over the life of the loan.

7. You are offered an investment with the following conditions:

- The cost of the investment is `$10,000`.
- The investment will pay `$2,000` at the end of year 1.
- The payout will grow at a rate of 5% per year for the next 4 years.

If your discount rate is 15%, calculate the net present value (NPV) of the investment and determine whether it is a good investment.
